#### StructType() Schema


```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
my_struct_schema = StructType([
                    StructField('Item_Identifier',StringType(),True),
                    StructField('Item_Weight',StringType(),True),
                    StructField('Item_Fat_Content',StringType(),True),
                    StructField('Item_Visibility',StringType(),True),
                    StructField('Item_MRP',StringType(),True),
                    StructField('Outlet_Identifier',StringType(),True),
                    StructField('Outlet_Establishment_Year',StringType(),True),
                    StructField('Outlet_Size',StringType(),True),
                    StructField('Outlet_Location_Type',StringType(),True),
                    StructField('Outlet_Type',StringType(),True),
                    StructField('Item_Outlet_Sales',StringType(),True)
])
```

```python
df = spark.read.format('csv')\
        .schema(my_struct_schema)\
        .option('header',True)\
        .load("/Volumes/workspace/default/new_volume/BigMart Sales.csv")
```

```python
df.printSchema()
```

#### union and Union byname

```python
data1 = [('1','khushi'),
         ('2','anushka')]
schema1 = 'id string, name string'
df1 = spark.createDataFrame(data1, schema1)

data2 = [('3','nisha'),
         ('4','shaurya')]
schema2 = 'id string, name string'
df2 = spark.createDataFrame(data2, schema2) 
```

```python
df1.display()
```

```python
df2.display()
```

##### union

```python
df1.union(df2).display()
```

##### union by name

```python
df1.unionByName(df2).display()
```

#### Date Functions

##### Current_Date

```python
df=df.withColumn('curr_date',current_date())
df.display()
```

#### date_add()

```python
df=df.withColumn('week_after',date_add('curr_date',7))
df.display()
```

##### date_sub()

```python
df.withColumn('week_before',date_sub('curr_date',7)).display()
```

##### datediff

```python
df.withColumn('datediff',datediff('curr_date','week_after')).display()
```

##### date_format

```python
df.withColumn('week_before',date_format('week_after','dd-MM-yyyy')).display()
```
