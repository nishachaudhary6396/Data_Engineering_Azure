## Task 1: Data Ingestion & Exploration
### Objective
#### Load the dataset into PySpark and perform exploratory analysis.

```python
import os

os.environ["HADOOP_HOME"] = "C:\\hadoop"
os.environ["PATH"] += ";C:\\hadoop\\bin"
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import when, count, col
```

```python
#create spark session
spark = SparkSession.builder\
.appName('Data Ingestion and Exploration')\
.getOrCreate()
```

```python
#load dataset
df = spark.read.csv("BNPParibas_Data.csv",
                    header=True,
                    inferSchema=True
                    )
```

```python
#Schema
df.printSchema()
```

```python
#  Record Count
df.count()
```

```python
# data Types
df.dtypes
```

```python
#null values
df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()
```

```python
#Duplicate Count
total_rows = df.count()
unique_rows = df.distinct().count()
Duplicate_Count = total_rows - unique_rows
Duplicate_Count
```

```python
#summary statistics
df.describe().show()
```
